### Cajero Automatico

```python
saldo = 2000.0
conectado = True

print("--- BIENVENIDO AL SISTEMA DE CAJERO INTELIGENTE ---")

while conectado == True:
    print("\nEscribe una acción: 'saldo', 'deposito', 'retiro' o 'salir'")
    comando = input("IA: Esperando comando... ")

    # CASO 1: Consulta de información (Estado de cuenta)
    if comando == "saldo":
        print("IA: Consultando base de datos...")
        print("Tu saldo actual es de: $" + str(saldo))
        break

    # CASO 2: Incremento de capital (Depósito)
    elif comando == "deposito":
        # Convierte la entrada de texto a un número decimal (float) [10]
        monto_in = float(input("IA: ¿Cuánto dinero vas a ingresar?: "))
        if monto_in > 0:
            saldo = saldo + monto_in # Realiza una suma aritmética para actualizar el saldo [11]
            print("IA: Dinero recibido. Tu nuevo saldo es: $" + str(saldo))
        else:
            print("IA ERROR: El monto debe ser mayor a cero.")
            break

    # CASO 3: Reducción de capital (Retiro con validación)
    elif comando == "retiro":
        monto_out = float(input("IA: ¿Cuánto dinero deseas retirar?: "))
        if monto_out > saldo:
            print("IA ERROR: No tienes suficiente dinero en tu cuenta.")
        else:
            saldo = saldo - monto_out
            print("IA: Retiro procesado. Toma tu efectivo.")
            print("Saldo restante: $" + str(saldo))
            break

    # CASO 4: Finalización del proceso (Salida)
    elif comando == "salir":
        print("IA: Cerrando conexión segura. ¡Gracias por usar el cajero!")
        conectado = False
        break
print("Elaborado por Alexandro Aguilar NL = 1")
```
